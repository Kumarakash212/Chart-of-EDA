Installs – Lower-bound install range (“10,000+”, “5,000,000+”); treat as ordinal.

Type – “Free” or “Paid”; one missing value.

Price – Price in USD with “$” prefix; “0” for free apps.

Content Rating – Age guidance: “Everyone”, “Teen”, “Mature 17+”, etc.

Genres – Comma-separated genre tags; may contain multiple values.

Last Updated – Date string (MMM DD, YYYY) when publisher last uploaded an APK.

Current Ver – Version string visible to user (e.g., “1.2.3”); 8 missing.

Android Ver – Minimum Android version required (e.g., “4.4 and up”); 3 missing.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# remove the missing value
data=data.dropna()

# remove duplicates values from dataset
data.drop_duplicates(inplace=True)

# convert data types
data['Last Updated'] = pd.to_datetime(data['Last Updated'],errors='coerce')

# fill the missing values
data['Rating']=data['Rating'].fillna(data['Rating'].mean())

#convert  into numeric
data['Reviews']=data['Reviews'].astype(int)

# Cleaning
data['Installs']=data['Installs'].astype(str).str.replace(',','', regex=False).str.replace('+','',regex=False).astype(int)
data['Size']=data['Size'].str.replace('M','',regex=False)
data['Price']=data['Price'].str.replace('$','',regex=False).astype(float)
data['Size']=pd.to_numeric(data['Size'],errors='coerce')

# Simplify
data['Genres']=data['Genres'].str.split(';').str[0]

# final
print(data.info())
print(data.describe)
print('Dataset is now ready for analysis')
````
